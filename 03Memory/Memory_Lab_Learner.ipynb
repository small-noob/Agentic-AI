### TODO 2 · `validate_record` — the write gate: three checks, five blanks

Decide whether ONE candidate record may enter the store. Return
`(True, "ok")` or `(False, "<short reason a human can act on>")`.

**No API call here, on purpose — this is an exam point.** The model produced
this record; the model does not also get to approve it. A gate whose verdict
can differ between runs cannot be reviewed and cannot be tested — the
attack-set cell below is exactly that test, and it needs no key.

**Credential shapes — provided, nothing to memorise.** The patterns in the
cell below cover this table:

| Credential | Shape | Example |
|---|---|---|
| OpenAI / Zhipu key | `sk-` + 8+ chars | `sk-proj-3f9Qd…` *(starter list)* |
| AWS access key | `AKIA` + 16 uppercase/digits | `AKIAIOSFODNN7EXAMPLE` *(starter list)* |
| GitHub token | `ghp_` + 20+ alphanumerics | `ghp_9zX8yW7vU6tS…` |
| Slack token | `xox` + one of `b a p r s` + `-` + 10+ chars | `xoxb-1234567890-…` |
| Assignment shape | `password= / passwd= / secret= / token= / api_key=` + a value | `password=hunter2` |

Three checks, in this order: **1. SECRETS** first, because a leak is the
only failure that can never be undone — scan the key and the value together;
**2. SHAPE** — real, non-placeholder strings; **3. ONE FACT PER RECORD** — a
record you cannot look up by one key is a record you cannot update.

**The blanks:** ① build the scan string — a gate that only scans the value
is a gate an attacker walks around by hiding the token in the key ·
② pattern hit · ③ value shape, mirroring the key check · ④ placeholder junk
(`"---"` is non-empty but says nothing — the attack set plants exactly this) ·
⑤ compound values.

And what the gate must **NOT** do: reject a record because the store already
holds it (that is TODO 3's NOOP), or holds that key with a different value
(its UPDATE). The gate asks *"is this record admissible?"* — TODO 3 asks
*"what does it mean, given what I already know?"*

````python
# TODO 2 — the patterns are provided (read them against the table above);
# fill the five blanks in the gate, then run. No API call — pure code.

SECRET_PATTERNS = memory_lab.SECRET_PATTERNS + [        # provided
    re.compile(r"\bghp_[A-Za-z0-9]{20,}"),              # GitHub personal access token
    re.compile(r"\bxox[baprs]-[A-Za-z0-9\-]{10,}"),     # Slack token
    re.compile(r"(?i)\b(password|passwd|secret|token|api_key)\s*[=:]\s*\S+"),  # assignment shape
]
COMPOUND_MARKERS = memory_lab.COMPOUND_MARKERS


def validate_record(self, record: dict, store) -> tuple[bool, str]:
    """Decide whether one candidate record may enter the store."""
    if not isinstance(record, dict):
        return False, "not a record"

    # 1. SECRETS — first, because a leak can never be undone. Key and value
    #    are scanned TOGETHER: models hide credentials in either.
    blob = ____                           # ①: build the one string to scan — key AND value together
    for pattern in SECRET_PATTERNS:
        if ____:                          # ②: did this pattern hit blob?
            return False, "secret"

    # 2. SHAPE — key and value must be real, non-placeholder strings.
    key = record.get("key")
    value = record.get("value")
    if not isinstance(key, str) or not key.strip():
        return False, "missing key"
    if ____:                              # ③: mirror the key check, for value
        return False, "missing value"
    if ____:                              # ④: not a single letter or digit in value?
        return False, "placeholder value"

    # 3. ONE FACT PER RECORD.
    if ____:                              # ⑤: any compound marker inside value?
        return False, "two facts in one record"

    # Deliberately NO comparison against store state: duplicates are
    # reconcile's NOOP, changed values its UPDATE.
    return True, "ok"


MemoryPolicy.validate_record = validate_record   # bind your answer
````

### TODO 3 · `reconcile` — two prompts, six blanks

Apply the accepted records to the store. All four verdicts occur in this
transcript, judged against the seed store:

| Verdict | Trigger in the transcript | Executes |
|---|---|---|
| ADD | "logs/access_2026-09.csv", "audit runs on day 1" | `store.add(record)` |
| UPDATE | "the fine is 250 now, not 200" | `store.supersede(key, record)` — old row → superseded |
| DELETE | "stop sending the report to security-team" | `store.soft_delete(key, source=...)` |
| NOOP | "the incident file is still incident_0812.txt" | `store.noop(key)` |

The split to hold onto: **the model judges the relationship, your code
performs the operation.** Judging needs language — `fine_per_violation` and
"the fine per record" are the same fact in different strings. Performing
must not: whether the store says 250 or 200 changes every total computed
later. Nothing is ever physically removed — UPDATE keeps the old row marked
superseded (the grader checks that row), revocation marks rather than erases.

**The prompts are the writing.** `RECONCILE_PROMPT` must define the four
verdicts, demand JSON only in a shape you specify (e.g. `{"verdict": "..."}`),
and say *judge by meaning, not string equality*. `REVOKE_PROMPT` must accept
**explicit revocations only**, say a changed value is an UPDATE, not a
revocation (conflate them and the fine gets deleted instead of updated), and
default to `{"revoked": []}` when in doubt.

**The blanks are the concepts:** ① look up what is stored · ② ③ the guards —
live models return UPDATE for keys that hold nothing; the code keeps
operations coherent with state · ④ ⑤ execute ADD / UPDATE · ⑥ the revocation
pass — "stop sending X" names a stored fact with **no new value**, so
extraction never produced a candidate for it; that is why this pass exists.

### TODO 4 · `build_context` — the ladder, seven blanks

Assemble the messages for ONE model call under `budget` tokens and return
`Assembled(messages, tokens, ladder)`. The budget is checked HERE, before
any `client.chat` would go out — checking usage afterwards is not a budget
check. Here it runs once, over the whole transcript, at a budget (600t) the
transcript does not fit into.

The skeleton hands you the structure; the blanks are the ladder's decisions:

| Stage | What (already in the skeleton) | Your blank |
|---|---|---|
| assemble | system prompt + the memory digest + the turns, measured | ② how does the digest enter the context? Part A's session 2 answered *because* of this line |
| **L0** | everything as-is | ① does it fit? |
| **L1** | free, lossless-ish: shorten oversized messages — they **stay** | ③ trim each message with `self.trim_oversized` |
| landed | a rung that fits returns immediately | ④ what does it return? |
| **L2 / L3** | one API call, lossy: the head becomes one note, the tail stays verbatim | ⑤ split head / tail — the verbatim tail is why the final instruction survives (a graded point) · ⑥ squeeze `head` with `self.compact` |
| still over | refuse to send — a budget merely reported is not a budget | ⑦ `raise ContextOverflow(used, budget, "<why>")` |

Why L1 before L2 (the grader checks the order): trimming is targeted, free,
and keeps the message; compaction costs an API call and loses detail across
every turn it touches. **Cheap before expensive, lossless before lossy.**
